```python

import numpy as np
import matplotlib.pyplot as plt
```

```python
# delimita el campo de trabajo en el eje x
N = 200
k_max = N // 2

x_min, x_max = -5*np.pi, 5*np.pi

x = np.linspace(x_min, x_max, N)
```

```python
# Define la función
f = np.exp(-x**2)
```

```python
def freq_axis(x):
    """Calcula los valores  k_j """
    N = len(x)
    k_max = N // 2
    if N % 2 == 0:
        k = [k_ for k_ in range(k_max)] + [k_ for k_ in range(-k_max, 0)]
    else:
        k = [k_ for k_ in range(k_max+1)] + [k_ for k_ in range(-k_max, 0)]    
    return np.array(k) * 2*np.pi / (x[-1] - x[0])
```

```python
# FFT transformada rápida de Fourier
c = np.fft.fft(f)
```

```python
# Calcula los valores k_j :
k = freq_axis(x)
```

```python
# Multiplica y realiza la transformada inversa de Fourier
F1 = np.fft.ifft(1j*k*c)
```

```python

# derivada exacta para comparación
f1 = -2*x*np.exp(-x**2)
```

```python
# Muestreo gráfico de la comparación precisa y calculo de la derivada
plt.plot(x, f1)
plt.plot(x, np.real(F1), 'o', ms=4)
plt.xlabel('$x$')
plt.ylabel('$f^{\,\prime}(x)$')
plt.grid()
```
